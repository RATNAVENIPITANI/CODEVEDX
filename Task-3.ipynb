```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("StudentsPerformance.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="gender", data=df)

plt.title("Number of Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["math score"], bins=20, edgecolor="black")

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
average_scores = [
    df["math score"].mean(),
    df["reading score"].mean(),
    df["writing score"].mean()
]

subjects = ["Math", "Reading", "Writing"]
```

```python
plt.figure(figsize=(7,5))

plt.bar(subjects, average_scores)

plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["math score"], df["reading score"])

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(x="gender", y="math score", data=df)

plt.title("Math Score Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Math Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x="test preparation course",
    y="math score",
    data=df
)

plt.title("Average Math Score by Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Math Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x="lunch",
    y="math score",
    data=df
)

plt.title("Average Math Score by Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Average Math Score")

plt.show()
```

```python
corr = df[
    ["math score", "reading score", "writing score"]
].corr()
```

```python
plt.figure(figsize=(7,5))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Student Scores")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df[
        ["math score", "reading score", "writing score"]
    ]
)

plt.title("Score Distribution by Subject")
plt.xlabel("Subject")
plt.ylabel("Score")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x="parental level of education",
    y="math score",
    data=df
)

plt.title("Average Math Score by Parental Education")
plt.xlabel("Parental Level of Education")
plt.ylabel("Average Math Score")

plt.xticks(rotation=45)

plt.show()
```
